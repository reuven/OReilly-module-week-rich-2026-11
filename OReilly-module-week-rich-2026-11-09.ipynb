# What's it for?

If I say `print('hello')`, it's going to come out as plain text on my terminal.

There is an option that I can use, namely to send "escape codes" to my terminal. That means literally sending the ESC character, followed by some very odd-looking characters often including `[` and `(`, which tell the terminal to turn on/off bold, italic, colors, etc.

Rich came around and decided to make it significantly easier to format our output.

```python
import rich
```

The most commonly used function in `rich` is `rich.print`, which is a full superset of the builtin Python `print` function. In fact, many people go so far as to use `from rich import print` -- which shadows the builtin `print` function, replacing it (sort of) with `rich.print`. (Really, we are defining a global variable which shadows the builtin variable.)

```python
rich.print('Hello out there!')
```

Rich's formatting typically uses `[]` at the start of whatever formatting you want, and then `[/ ]` when you want to stop the formatting. So, for example, if you want red text, you can say `'[red]text[/red]'` 

```python
rich.print('Hello [red]out[/red] there!')
```
